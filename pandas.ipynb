```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

print(amounts)
print("합계", sum(amounts))

print(amounts * 2) # 각항에 곱하기 2가 아닌 두번 반복
print("==="*3)

```

```python
import numpy as np

arr = np.array(amounts) # amount를 숫자 배열로 바꿈

print(arr * 2)
# [ 6400  3000 30000 13000 64000  9600 21800  3000  5400  6000 37000 14000]
print("합계:", arr.sum())
# 합계: 106600
print("평균:", round(arr.mean()))
# 평균: 8883
print("최소:", arr.min())
# 최소: 1500
print("최대:", arr.max())
# 최대: 32000

print("모양:", arr.shape, " 데이터 타입: ", arr.dtype)
# 모양: (12,)  데이터 타입:  int64 : 1차원, 0이 64개
```

```python
print(arr >= 10000)
# [False False  True False  True False  True False False False  True False]: amount의 >= 10000 True False 값
arr[arr >= 10000]
# array([15000, 32000, 10900, 18500])
```

```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

print(shops)


food =0
# zip : 같은 요소끼리 묶
for c, a in zip(categories, amounts):
    if c == "식비":
        food = food + a
# 식비를 다 더한 값인데 해석이 어려움
print(food)
# 24200
```

```python
import pandas as pd

# DataFrame의 'DF'는 대문자 DataFrame 이란?
spend = pd.DataFrame({
    "day": [2, 3, 4, 5, 7, 9, 12 , 14 ,16, 19, 20, 25],
    "shop": shops,
    "category": categories,
    "amount": amounts
})

spend.head(5) # 5줄만 출력

# spend.info() 
# sql에서 \d와 같음 컬럼의 속성을 알수있다

# print(spend.shape)
# # (12, 4) 이차원 배열: 가로 세로 있는 표?
```

```python
spend["amount"]
print("합계:", spend["amount"].sum())
print("평균:", round(spend["amount"].mean(),2))
```

```python
# spend["shop", "amounts"] "shop, amounts"

# spend[컬럼명]
# spend[[컬럼1, 컬럼2, ...]

spend[ ["shop", "amount"] ]
```

### 조건 설정


```python
# 조건식 예시
spend["category"]
```

```python
# spend[조건] -> 조건이 True인 행만 리턴
# 조건식 예시 spend["category"] == "문화"

# spend["컬럼명"] -> "컬럼1"의 로우 모두 리턴

spend[spend["category"] == "문화"]
```

```python
# 조건식 spend["amount"] >= 10_000
spend[spend["amount"] >= 10_000]
```

```python
# 식비 5000>= 사용한 행을 출력
# & 사용시 () 사용
spend[ (spend["category"] == "식비") & (spend["amount"] >= 5000) ]
```

```python
# 오름차순 : sort_values()

# 내림차순 : 뒤에 , ascending=False 추가
top3 = spend.sort_values("amount", ascending=False).head(3)
```

```python
# spend["새 컬럼 생성"] = 계산식
spend["point"] = spend["amount"] // 100 # 100원 마다 1포인트,  32: 3200//100 (몫) 3200 % 100 (나머지)
spend.head(5)
print(spend["point"].sum())


spend
```

```python
# 특정값 찾아서 바꾸기
spend.loc[2, "amount"] = 14000
# point가 amount의 가변이나 바뀌지않아 새로 덮어씌움
spend["point"] = spend["amount"] // 100
spend.head(3)
```

```python
# 그룹별로 더한 값
spend.groupby("category")["amount"].sum()
```

```python
# 그룹별로 더한 값 내림차순
spend.groupby("category")["amount"].sum().sort_values(ascending=False)
```
